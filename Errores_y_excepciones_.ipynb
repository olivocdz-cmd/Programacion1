Excepciones y errores en Python

```python
# Código 1
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()
    password = input("Contraseña: ")

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    if usuarios[usuario] != password:
        raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

```python
# Código 2
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    elif usuarios[usuario] != input("Contraseña: "):
      raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

La diferencia entre el primer código y el segundo radica en el comportamiento del error cuando el usuaario es incorrecto. En el primer caso, como el «try» está anclado al usuario y a la contraseña, si hay un error en uno de los dos, intentará ejecutar el otro. De tal manera que sólo salta el error de usuario o de contraseña al final. En el segundo caso, como el «try» sólo se imparte sobre el ususario, cuando una de la condición establecida (de que debe estar en en diccionario) falla, salta el error.

```python
productos = {
    "pan": 1.5,
    "leche": 2.0,
    "huevos": 3.0
}

try:

  print("Productos disponibles:", productos)
  nombre = input("¿Qué producto quieres?: ").lower()

  if nombre not in productos:
      raise KeyError

  cantidad = int(input("Cantidad: "))
  if cantidad <= 0:
    raise ValueError

  total = productos[nombre] * cantidad
  print("Total a pagar:", total)


except ValueError:
  print("Error: el número debe ser entero")
except KeyError:
  print("El producto no está disponible")
```

**Introducción al manejo de excepciones regulares**:

Una expresión regular (o regex, del inglés regular expression) es una secuencia de caracteres que forma un patrón de búsqueda. Se utiliza para determinar si una cadena de texto cumple con ese patrón, o para extraer, reemplazar o dividir partes de ella.

En Python se usa el módulo re

```python
import re
texto = "Python es un lenguaje"
resuelto = re.search("Python", texto)
print(resuelto)
```

```python
resultado1 = re.search("Java", texto)
print(resultado1)
```

re.findall(patrón, texto) – Devuelve todas las coincidencias en una lista.

Con re.findall(r"\d", texto), podemos filtar llos dígitos y con re.findall(r"\d+", texto), sirve para extraer todas las secuencias de uno o más dígitos (números enteros) que aparezcan en la cadena texto, y las devuelve en una lista. Por otro lado, re.findall(r"\w+", texto) busca todas las palabras dentro de texto y las guarda en una lista. Tambiñen se pueden fltrar letras específicas como las vocales, por ejemplo, en re.findall(r"[aeiou]", texto); las letras mayúsculas con re.findall(r"[A-Z]", texto) o, incluso, caracteres concreto como los espacios con re.findall(r"\s", texto).
